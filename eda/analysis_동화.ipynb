```python
kpli_mean_dp_day_dt = kpli_df.groupby([
    '도착항코드(DEST_PORT_CODE)','기준일자(DATE)']).agg(
    항만효율성=('항만효율성(PORT_EFFICIENCY)', 'mean'),
    총항해시간=('총항해시간(TOTAL_SAILING_TIME)', 'mean'),
    대기시간=('대기시간(WAITING_TIME)', 'mean'),
    항만정시성=('항만정시성(ON_TIME_PERFORMANCE)', lambda x: x.abs().mean()),
).reset_index()

kpli_mean_dp_mon_dt = kpli_mean_dp_day_dt.groupby([
    '도착항코드(DEST_PORT_CODE)',
    pd.to_datetime(kpli_mean_dp_day_dt['기준일자(DATE)']).dt.to_period('M')
])[['항만효율성',
 '총항해시간',
 '대기시간',
 '항만정시성']].mean().reset_index()

kpli_mean_dmon_dt = kpli_mean_dp_mon_dt.groupby(['기준일자(DATE)']
)[['항만효율성',
 '총항해시간',
 '대기시간',
 '항만정시성']].mean().reset_index()

kpli_mean_dmon_dt
```

```python

for port in country_df['도착항명(DEST_PORT_NAME)'].unique():

    port_df = country_df[
        country_df['도착항명(DEST_PORT_NAME)'] == port
    ]

    q1 = port_df['대기시간(WAITING_TIME)'].quantile(0.25)
    q3 = port_df['대기시간(WAITING_TIME)'].quantile(0.75)

    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    # port_df['lower'] = lower
    # port_df['upper'] = upper
    print(port, lower, upper)
```

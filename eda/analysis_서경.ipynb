```python
#1
TARGET_PATH = '../data/data.csv'
TARGET_PATH3 = '../data/data3.csv'
```

```python
#2
df = pd.read_csv(TARGET_PATH, encoding='utf-8-sig')
df3 = pd.read_csv(TARGET_PATH3, encoding='cp949')
```

```python
#3
region_map = {
    '상하이항': '극동아시아 지역',
    '칭다오항': '극동아시아 지역',
    '홍콩항': '극동아시아 지역',

    '싱가포르항': '동남아시아 지역',
    '램차방항': '동남아시아 지역',
    '호치민항': '동남아시아 지역',

    '로스앤젤레스항': '북미주 지역',
    '뉴욕항': '북미주 지역',

    '함부르크항': '유럽 지역',
    '로테르담항': '유럽 지역',
    '안트베르펜항': '유럽 지역',

    '제벨알리항': '중동 지역',

    'KOBC PORT&LOGISTICS INDICATOR (종합)': '종합'
}

df['지역'] = df['도착항명(DEST_PORT_NAME)'].map(region_map)
```

```python
df3['지역'].unique()
```

```python
#5
# 날짜 형식 변환
df['기준일자(DATE)'] = pd.to_datetime(df['기준일자(DATE)'])

# 연월 컬럼 만들기
df['연월'] = df['기준일자(DATE)'].dt.strftime('%Y년 %m월')
```

```python
df.head()
```

```python
print(df.columns.tolist())
```

```python
#6
regional = df.groupby(['지역','연월'])['항만효율성(PORT_EFFICIENCY)'].mean().round(2).reset_index()

# 5월 데이터 빼기
regional = regional[regional['연월'] != '2025년 05월']
```

```python
#7
regional = regional[regional['지역'] != '종합']
```

```python
regional['지역'].unique()
```

```python
volume_long = df3.melt(
    id_vars='지역',
    var_name='연월',
    value_name='물동량'
)
```

```python
merged = pd.merge(
    regional,
    volume_long,
    on=['지역', '연월'],
    how='left'
)
```

```python
merged.head()
```

```python
# 종합...

fig = make_subplots(
    specs=[[{'secondary_y': True}]]
)

# 1. 지역별 물동량 → 누적 막대
for region in merged['지역'].unique():

    temp = merged[merged['지역'] == region]

    fig.add_trace(
        go.Bar(
            x=temp['연월'],
            y=temp['물동량'],
            name=f'{region} 물동량',
            legendgroup=region
        ),
        secondary_y=False
    )


# 2. 지역별 항만효율성 → 선
for region in merged['지역'].unique():

    temp = merged[merged['지역'] == region]

    fig.add_trace(
        go.Scatter(
            x=temp['연월'],
            y=temp['항만효율성(PORT_EFFICIENCY)'],
            mode='lines+markers',
            name=f'{region} 효율성',
            legendgroup=region
        ),
        secondary_y=True
    )


# 막대 누적
fig.update_layout(
    barmode='group',
    width=1100,
    height=600
)

# 왼쪽 y축
fig.update_yaxes(
    title_text='물동량',
    secondary_y=False
)

# 오른쪽 y축
fig.update_yaxes(
    title_text='항만효율성',
    secondary_y=True
)

fig.update_xaxes(
    title_text='연월'
)

fig.show()
```

```python
def region_graph(region):

    # 선택한 지역만 가져오기
    temp = merged[merged['지역'] == region]

    fig = make_subplots(
        specs=[[{'secondary_y': True}]]
    )

    # 물동량 막대그래프
    fig.add_trace(
        go.Bar(
            x=temp['연월'],
            y=temp['물동량'],
            name='물동량'
        ),
        secondary_y=False
    )

    # 항만효율성 선그래프
    fig.add_trace(
        go.Scatter(
            x=temp['연월'],
            y=temp['항만효율성(PORT_EFFICIENCY)'],
            mode='lines+markers',
            name='항만효율성'
        ),
        secondary_y=True
    )

    fig.update_layout(
        width=1100,
        height=600
    )

    fig.update_yaxes(
        title_text='물동량',
        secondary_y=False
    )

    fig.update_yaxes(
        title_text='항만효율성',
        secondary_y=True
    )

    fig.update_xaxes(
        title_text='연월'
    )

    return fig
```
